## Zou-He BCs

Given the velocity $\vec{u} = (u, v)$, the *macroscopic velocity* now splits into two components
$$
\begin{aligned}
\rho u &= \sum_i cf_ie_{ix}, \\
&= f_1 - f_3 + f_5 - f_6 - f_7 + f_8, \\
f_1 + f_5 + f_8 &= \rho u + (f_3 + f_6 + f_7); \\\\
\rho v &= \sum_i cf_ie_{iy}, \\
&= f_2 - f_4 + f_5 + f_6 - f_7 - f_8, \\
f_5 - f_8 &= \rho v - f_2 + f_4 - f_6 + f_7.
\end{aligned}
$$

With
$$
f_1 - f_1^{\text{eq}} = f_3 - f_3^{\text{eq}} \implies f_1 = f_3 + \left(f_1^{\text{eq}} - f_3^{\text{eq}}\right),
$$
the D2Q9 model now becomes
$$
f_i^{\text{eq}} = w_i\rho\left[1 + 3(\vec{e}_i \cdot \vec{u}) + \frac{9}{2}(\vec{e}_i \cdot \vec{u})^2 - \frac{3}{2}\vec{u}^2\right].
$$
Since $\vec{e}_1 = (1, 0)$, $\vec{e}_3 = (-1, 0)$, and $w_1 = w_3 = \frac{1}{9}$, we obtain
$$
\begin{aligned}
f_1^{\text{eq}} - f_3^{\text{eq}} &= \frac{1}{9}\rho\left[3u - (-3u)\right], \\
&= \frac{2}{3}\rho u, \\
\implies f_1 &= f_3 + \frac{2}{3}\rho u.
\end{aligned}
$$
Substitute $f_1 = f_3 + \frac{2}{3}\rho u$, we have
$$
f_5 + f_8 = f_6 + f_7 + \frac{2}{3}\rho u.
$$
Thus, we can get $f_5$ with
$$
\begin{aligned}
f_5 &= \frac{1}{2}\left[(f_5 + f_8) + (f_5 - f_8)\right], \\
&= f_7 - \frac{1}{2}(f_2 - f_4) + \frac{1}{6}\rho u + \frac{1}{6}\rho v.
\end{aligned}
$$
Similarly, $f_8$ is
$$
\begin{aligned}
f_8 &= \frac{1}{2}\left[(f_5 + f_8) - (f_5 - f_8)\right], \\
&= f_6 + \frac{1}{2}(f_2 - f_4) + \frac{1}{6}\rho u - \frac{1}{6}\rho v.
\end{aligned}
$$

# Steady Plane Poiseuille Flow

The incompressible Navier-Stokes equations are
$$
\begin{aligned}
\nabla \cdot \vec{u} &= 0; \\
\rho\left(\frac{\partial \vec{u}}{\partial t} + \vec{u} \cdot \nabla \vec{u}\right) &= -\nabla p + \mu\nabla^2\vec{u}.
\end{aligned}
$$
We assume the steady plance Poiseuille flow only in $x$-direction, and varies only with $y$. That is,
$$
\vec{u} = \left(u(y), 0\right).
$$
Thus,
$$
\begin{aligned}
\frac{\partial u}{\partial t} &= 0; \\
u\frac{\partial u}{\partial x} &= 0; \\
v\frac{\partial u}{\partial y} &= 0.
\end{aligned}
$$
This is due to the steady flow, $u$ has no $x$-dependence, and $v = 0$. The $x$-momentum equation becomes
$$
0 = -\frac{\partial p}{\partial x} + \mu\frac{\partial^2u}{\partial y^2} \implies \mu\frac{\partial^2u}{\partial y^2} = \frac{\partial p}{\partial x},
$$
where the pressure gradient is
$$
\frac{\partial p}{\partial x} = \frac{P_1 - P_0}{L}.
$$

Solving for the differential equation, we obtain
$$
\frac{\mathrm{d}^2u}{\mathrm{d}y^2} = \frac{1}{\mu}\frac{P_1 - P_0}{L}.
$$
After one integration, we have
$$
\frac{\mathrm{d}u}{\mathrm{d}y} = \frac{P_1 - P_0}{\mu L}y + C_1.
$$
After another integration, we get
$$
u(y) = \frac{P_1 - P_0}{2\mu L}y^2 + C_1y + C_2.
$$
Apply the boundary conditions, the coefficients are
$$
\begin{aligned}
u(0) = 0 &\implies C_2 = 0; \\
u(H) = 0 &\implies \frac{P_1 - P_0}{2\mu L}H^2 + C_1H = 0; \\
&\implies C_1 = -\frac{P_1 - P_0}{2\mu L}H.
\end{aligned}
$$
As a result, the solution is
$$
u(y) = \frac{\Delta p}{2\mu L}y(y - H).
$$

The criterion for steady state is given by
$$
\frac{\sum_{ij}\left|u_{ij}^{n + 1} - u_{ij}^n\right|}{\sum_{ij}\left|u_{ij}^{n + 1}\right|} \le 5.0 \times 10^{-9}.
$$
The flow field will be no longer evolving in time if the relative change between the two consecutive iterations becomes extremely small.
